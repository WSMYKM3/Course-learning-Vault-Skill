```python
%matplotlib inline
import numpy as np
import pandas as pd
from math import exp
import matplotlib.pyplot as plt
from sklearn.preprocessing import normalize
from sklearn.metrics import accuracy_score
```

### Reading the dataset

```python
data_filename = 'insurance_claim.csv'
df = pd.read_csv(data_filename)
```

### Assigning predictor and response variables

```python
x = df['age'].values
y = df['insuranceclaim']
```

```python
plt.plot(df[['age']],y,'o', markersize=7,color="#011DAD")
plt.xticks(np.arange(18, 80, 4.0))
plt.xlabel("Age")
plt.ylabel("Insurance claim")
plt.yticks((0,1), labels=('No', 'Yes'))
plt.show()
```

### Beta Estimation

```python
beta0 = -40

beta1 = 1
```

### Predicting the response

```python
# Define the function to calculate the prediction of x using beta0 and beta1
logit = []
for i in x:
    beta = -1*(beta0 + (beta1 * i))
    logit.append(1/(1+exp(beta)))

y_pred = [1 if x >= 0.5 else 0 for x in logit]
```

### Computing the accuracy

```python
accuracy = accuracy_score(y, y_pred)
print (accuracy)
```

### Plotting the graph

```python
plt.plot(df[['age']],y,'o', markersize=7,color="#011DAD")
plt.plot(df[['age']],logit,linewidth=2,color='black')
plt.xticks(np.arange(18, 80, 4.0))
plt.xlabel("Age")
plt.ylabel("Insurance claim")
plt.yticks((0,1), labels=('No', 'Yes'))
plt.show()
```
